```python
# Import pandas
import pandas as pd
```

```python
# Load the original dataset
file_path = "../data/raw/dataset-tickets-multi-lang-4-20k.csv"

df = pd.read_csv(file_path)

print("Original shape:", df.shape)
```

```python
# Keep only columns needed for prompt experiments
columns_to_keep = [
    "subject",
    "body",
    "answer",
    "type",
    "queue",
    "priority",
    "language"
]

df_clean = df[columns_to_keep].copy()

df_clean.head()
```

```python
# Keep only columns needed for prompt experiments
columns_to_keep = [
    "subject",
    "body",
    "answer",
    "type",
    "queue",
    "priority",
    "language"
]

df_clean = df[columns_to_keep].copy()

df_clean.head()
```

```python
# Keep only English customer support tickets
df_clean = df_clean[df_clean["language"] == "en"].copy()

print("English tickets:", len(df_clean))
```

```python
# Check missing values after filtering
df_clean.isnull().sum()
```

```python
# Fill missing subjects with a placeholder
df_clean["subject"] = df_clean["subject"].fillna("No subject")

# Remove rows missing important text fields
df_clean = df_clean.dropna(subset=["body", "answer"])

print("Rows after handling missing values:", len(df_clean))
```

```python
# Remove unnecessary leading and trailing spaces
text_columns = ["subject", "body", "answer"]

for col in text_columns:
    df_clean[col] = df_clean[col].str.strip()
```

```python
# Clean whitespace in label columns
category_columns = ["type", "queue", "priority", "language"]

for col in category_columns:
    df_clean[col] = df_clean[col].str.strip()
```

```python
# Remove duplicate records if any exist
before = len(df_clean)

df_clean = df_clean.drop_duplicates()

after = len(df_clean)

print("Duplicates removed:", before - after)
```

```python
# Reset row index after cleaning
df_clean = df_clean.reset_index(drop=True)

df_clean.head()
```

```python
# Display final dataset information
print("Final shape:", df_clean.shape)

print("\nMissing values:")
print(df_clean.isnull().sum())

print("\nLanguages:")
print(df_clean["language"].value_counts())
```

```python
# Check labels used later for classification
print("Ticket Types:")
print(df_clean["type"].value_counts())

print("\nPriorities:")
print(df_clean["priority"].value_counts())

print("\nQueues:")
print(df_clean["queue"].value_counts())
```

```python
# Save the cleaned dataset
output_path = "../data/processed/cleaned_customer_support_tickets.csv"

df_clean.to_csv(output_path, index=False)

print("Cleaned dataset saved successfully.")
```
