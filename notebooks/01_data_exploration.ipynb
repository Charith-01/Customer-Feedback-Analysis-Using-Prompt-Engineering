```python
# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Load the raw customer support ticket dataset
file_path = "../data/raw/dataset-tickets-multi-lang-4-20k.csv"

df = pd.read_csv(file_path)
```

```python
# Display number of rows and columns
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
# Count missing values in each column
df.isnull().sum()
```

```python
# Count duplicate records
print("Duplicate rows:", df.duplicated().sum())
```

```python
# Display dataset structure and memory usage
df.info()
```

```python
# Check unique values in categorical columns
for col in ["type", "queue", "priority", "language"]:
    if col in df.columns:
        print(f"\n{col.upper()}")
        print(df[col].value_counts())
```

```python
# Preview important text columns
text_cols = ["subject", "body", "answer"]

for col in text_cols:
    if col in df.columns:
        print(f"\n--- {col.upper()} ---")
        print(df[col].dropna().head(3))
```

```python
# Calculate character length of text columns
for col in ["subject", "body", "answer"]:
    if col in df.columns:
        df[f"{col}_length"] = df[col].fillna("").str.len()

df[["subject_length", "body_length", "answer_length"]].describe()
```
