```python
# Import pandas
import pandas as pd
```

```python
# Load the original dataset
file_path = "../data/raw/dataset-tickets-multi-lang-4-20k.csv"

df = pd.read_csv(file_path)

print("Original shape:", df.shape)
```

```python
# Keep only columns needed for prompt experiments
columns_to_keep = [
    "subject",
    "body",
    "answer",
    "type",
    "queue",
    "priority",
    "language"
]

df_clean = df[columns_to_keep].copy()

df_clean.head()
```

```python
# Keep only columns needed for prompt experiments
columns_to_keep = [
    "subject",
    "body",
    "answer",
    "type",
    "queue",
    "priority",
    "language"
]

df_clean = df[columns_to_keep].copy()

df_clean.head()
```

```python
# Keep only English customer support tickets
df_clean = df_clean[df_clean["language"] == "en"].copy()

print("English tickets:", len(df_clean))
```

```python
# Check missing values after filtering
df_clean.isnull().sum()
```

```python
# Fill missing subjects with a placeholder
df_clean["subject"] = df_clean["subject"].fillna("No subject")

# Remove rows missing important text fields
df_clean = df_clean.dropna(subset=["body", "answer"])

print("Rows after handling missing values:", len(df_clean))
```

```python
# Remove unnecessary leading and trailing spaces
text_columns = ["subject", "body", "answer"]

for col in text_columns:
    df_clean[col] = df_clean[col].str.strip()
```

```python
# Clean whitespace in label columns
category_columns = ["type", "queue", "priority", "language"]

for col in category_columns:
    df_clean[col] = df_clean[col].str.strip()
```

```python
# Remove duplicate records if any exist
before = len(df_clean)

df_clean = df_clean.drop_duplicates()

after = len(df_clean)

print("Duplicates removed:", before - after)
```

```python
# Reset row index after cleaning
df_clean = df_clean.reset_index(drop=True)

df_clean.head()
```

```python
# Display final dataset information
print("Final shape:", df_clean.shape)

print("\nMissing values:")
print(df_clean.isnull().sum())

print("\nLanguages:")
print(df_clean["language"].value_counts())
```

```python
# Check labels used later for classification
print("Ticket Types:")
print(df_clean["type"].value_counts())

print("\nPriorities:")
print(df_clean["priority"].value_counts())

print("\nQueues:")
print(df_clean["queue"].value_counts())
```

```python
# Save the cleaned dataset
output_path = "../data/processed/cleaned_customer_support_tickets.csv"

df_clean.to_csv(output_path, index=False)

print("Cleaned dataset saved successfully.")
```

```python
# Load the cleaned customer support dataset
clean_df = pd.read_csv(
    "../data/processed/cleaned_customer_support_tickets.csv"
)

clean_df.shape
```

```python
# Sample 30 tickets from each ticket type
experiment_df = (
    clean_df
    .groupby("type", group_keys=False)
    .apply(lambda x: x.sample(n=30, random_state=42))
    .reset_index(drop=True)
)

print("Experiment dataset shape:", experiment_df.shape)
```

```python
# Create a balanced sample of 30 tickets from each ticket type
experiment_df = (
    clean_df
    .groupby("type", group_keys=False)
    .sample(n=30, random_state=42)
    .reset_index(drop=True)
)

print("Experiment dataset shape:", experiment_df.shape)
```

```python
# Verify ticket type distribution
experiment_df["type"].value_counts()
```

```python
# Inspect priority distribution in the sample
experiment_df["priority"].value_counts()
```

```python
# Add a unique experiment ID
experiment_df.insert(
    0,
    "ticket_id",
    range(1, len(experiment_df) + 1)
)

experiment_df.head()
```

```python
# Shuffle rows so ticket types are mixed
experiment_df = experiment_df.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)
```

```python
# Recreate sequential ticket IDs after shuffling
experiment_df["ticket_id"] = range(1, len(experiment_df) + 1)
```

```python
# Preview selected experiment records
experiment_df.head(10)
```

```python
# Save the balanced dataset used for prompt experiments
experiment_path = "../data/processed/prompt_experiment_tickets.csv"

experiment_df.to_csv(
    experiment_path,
    index=False
)

print("Prompt experiment dataset saved successfully.")
```
